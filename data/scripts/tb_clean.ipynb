```python
import pandas as pd
```

```python
tb = pd.read_csv("../processed/TB/data.csv")
```

```python
tb = tb[tb["image_type"] != "sick_but_no_tb"].reset_index(drop=True)
```

```python
tb
```

```python
tb.rename(columns={"fname": "file_name", "image_height": "height", "image_width": "width", "image_type": "classification"}, inplace=True)
```

```python
tb = tb[["file_name", "width", "height", "classification", "bbox"]]
```

```python
# Replace values in tb['classification']: 'healthy' to 'Normal', 'tb' to 'Tuberculosis'
tb["classification"] = tb["classification"].replace({"healthy": "Normal", "tb": "Tuberculosis"})
tb["classification"].unique()
```

```python
tb.to_csv("../processed/TB/metadata.csv", index=False)
```
