```python
import pandas as pd
```

```python
covid = pd.read_excel("../raw/COVID-19 Radiography Database/COVID.metadata.xlsx")
normal = pd.read_excel("../raw/COVID-19 Radiography Database/Normal.metadata.xlsx")
pneumonia = pd.read_excel("../raw/COVID-19 Radiography Database/Viral Pneumonia.metadata.xlsx")
```

```python
covid["classification"] = "COVID-19"
normal["classification"] = "Normal"
pneumonia["classification"] = "Pneumonia"
```

```python
# Combine normal and covid DataFrames into one
df_combined = pd.concat([normal, covid], ignore_index=True)
df_combined.to_csv("../processed/Covid_19/metadata.csv", index=False)
```

```python
df = pd.read_csv('../processed/Covid_19/metadata.csv')
```

```python
df.head()
```

```python
# Create 'file_name' column by combining 'FILE NAME' and 'FORMAT' (lowercase) with a dot
if 'FILE NAME' in df.columns and 'FORMAT' in df.columns:
    df['file_name'] = df['FILE NAME'].astype(str) + '.' + df['FORMAT'].str.lower()
    print(df[['FILE NAME', 'FORMAT', 'file_name']].head())
else:
    print("df does not have the required columns 'FILE NAME' and 'FORMAT'.")

# Split 'SIZE' column into 'width' and 'height'
if 'SIZE' in df.columns:
    df[['width', 'height']] = df['SIZE'].str.split('*', expand=True).astype(int)
    print(df[['SIZE', 'width', 'height']].head())
else:
    print("df does not have a 'SIZE' column.")
```

```python
df = df[["file_name", "width", "height", "classification"]]
```

```python
df.to_csv('../processed/Covid_19/metadata.csv', index=False)
```
